```python
df["date_time"] = pd.to_datetime(df["date_time"])
df["year"] = df["date_time"].dt.year
df["month"] = df["date_time"].dt.month
df["day"] = df["date_time"].dt.day
df["hour"] = df["date_time"].dt.hour
```

```python
hourly_traffic = df.groupby('hour')['traffic_volume'].mean()
hourly_traffic
```

```python
print(hourly_traffic.idxmax())
print(hourly_traffic.idxmin())
```

```python
day_traffic = df.groupby('day')['traffic_volume'].mean()
day_traffic
```

```python
df['day_of_week'] = df['date_time'].dt.day_name()
df[['date_time', 'day_of_week']].head(10)
```

```python
day_traffic = df.groupby('day_of_week')['traffic_volume'].mean()
day_traffic
```

```python
day_traffic.idxmax()
```

```python
day_traffic.idxmin()
```

```python
day_order = [
    'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'
]
df['day_of_week'] = pd.Categorical(df['day_of_week'], categories=day_order, ordered=True)
```

```python
day_traffic = df.groupby('day_of_week', observed=True)['traffic_volume'].mean()
day_traffic
```

```python
df['is_weekend'] = df['day_of_week'].isin(['Saturday', 'Sunday'])
df[['day_of_week', 'is_weekend']].head(10)
```

```python
weekend_traffic = df.groupby('is_weekend', observed=True)['traffic_volume'].mean()
weekend_traffic
```

```python
import matplotlib.pyplot as plt

weekend_traffic.plot(kind='bar')

plt.title('Average traffic: Weekdays vs Weekend')
plt.xlabel('Weekend')
plt.ylabel('Average traffic')
plt.xticks([0, 1], ['Weekdays', 'Weekend'], rotation=0)
plt.show()
```
