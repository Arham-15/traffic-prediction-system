### Traffic Volume Distribution

The traffic volume distribution is not normally distributed and shows multiple peaks. 
This is expected because traffic varies considerably across different hours, days, and 
other conditions.

The boxplot shows traffic volume ranging from approximately 0 to 7,300, with a median 
around 3,400. No extreme outliers are clearly visible from the boxplot.

```python
plt.figure(figsize=(12, 5))

hourly_traffic.plot(kind='line', marker='o')

plt.title('Average Traffic Volume by Hour')
plt.xlabel('Hour of Day')
plt.ylabel('Average Traffic Volume')
plt.xticks(range(24))
plt.grid()

plt.show()
```

### Traffic Volume by Hour

Average traffic volume varies substantially throughout the day. Traffic is lowest
during the early morning hours and increases sharply from around 04:00 onward.

Traffic remains high during the daytime and reaches its highest average around
16:00. After 17:00, traffic volume gradually decreases toward the night.

This shows that the time of day is an important feature for predicting traffic volume.

```python
hour_weekend = df.groupby(['is_weekend', 'hour'],observed=True)['traffic_volume'].mean()
hour_weekend
```

```python
plt.figure(figsize=(12, 5))

for weekend, data in hour_weekend.groupby(level=0):
    label = 'Weekend' if weekend else 'Weekday'
    plt.plot(
        data.index.get_level_values('hour'),
        data.values,
        marker='o',
        label=label
    )

plt.title('Average Traffic Volume by Hour: Weekday vs Weekend')
plt.xlabel('Hour of Day')
plt.ylabel('Average Traffic Volume')
plt.xticks(range(24))
plt.legend()
plt.grid()

plt.show()
```

### Hourly Traffic: Weekday vs Weekend

The hourly traffic pattern differs between weekdays and weekends. Weekdays show
stronger morning and afternoon peaks, while weekends show a more gradual increase
in traffic during the morning and a relatively stable daytime pattern.

This indicates that the interaction between time of day and whether a day is a
weekday or weekend may be useful for traffic prediction.

```python
monthly_traffic = df.groupby('month')['traffic_volume'].mean()
```

```python
plt.figure(figsize=(10, 5))

monthly_traffic.plot(kind='bar')

plt.title('Average Traffic Volume by Month')
plt.xlabel('Month')
plt.ylabel('Average Traffic Volume')
plt.xticks(rotation=0)

plt.show()
```

### Traffic Volume by Month

Average traffic volume varies moderately across the months. June has the highest
average traffic volume, while December has the lowest in this dataset.

The monthly variation is relatively small compared with the much stronger hourly
variation. Therefore, month may provide some seasonal information for the model,
while hour appears to capture a stronger traffic pattern.

```python
weather_traffic = df.groupby('weather_main')['traffic_volume'].mean().sort_values(ascending=False)
```

```python
plt.figure(figsize=(10, 5))

weather_traffic.plot(kind='bar')

plt.title('Average Traffic Volume by Weather')
plt.xlabel('Weather')
plt.ylabel('Average Traffic Volume')
plt.xticks(rotation=45)

plt.show()
```

### Traffic Volume by Weather

Average traffic volume varies across weather conditions. Cloudy and hazy conditions
have relatively high average traffic volumes, while squall has the lowest average
in this dataset.

These are descriptive relationships and do not by themselves establish that weather
causes changes in traffic volume.

```python
corr = df[
    [
        'temp',
        'rain_1h',
        'snow_1h',
        'clouds_all',
        'year',
        'month',
        'day',
        'hour',
        'traffic_volume'
    ]
].corr()

corr
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(corr, cmap='coolwarm', fmt='.2f', annot=True)

plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title('Correlation Matrix')
plt.show()
```

### Correlation Analysis

The correlation matrix shows that hour has the strongest linear correlation with
traffic volume (approximately 0.35). Temperature has a weaker positive correlation
(approximately 0.13), while rain, snow, clouds, year, month, and day show very weak
linear correlations with traffic volume.

However, correlation only measures linear relationships. The hourly analysis showed
a much more pronounced non-linear pattern, so weak correlation does not necessarily
mean that a feature has no predictive value.

```python
yearly_traffic = df.groupby('year')['traffic_volume'].mean()

yearly_traffic
```

```python
plt.figure(figsize=(10, 5))

yearly_traffic.plot(kind='line', marker='o')

plt.title('Average Traffic Volume by Year')
plt.xlabel('Year')
plt.ylabel('Average Traffic Volume')
plt.xticks(yearly_traffic.index)

plt.grid()
plt.show()
```

### Traffic Volume by Year

Average traffic volume varies across the years in the dataset. The average is
highest in 2017 and lowest in 2016.

The variation is moderate, while the hourly analysis showed a much stronger
pattern. Year may still provide useful temporal information to the model.

```python
time_diff = df['date_time'].sort_values().diff()

time_diff.value_counts().head(10)
```

```python
time_diff[time_diff > pd.Timedelta(hours=1)].head(20)
```

```python
gaps = time_diff[time_diff > pd.Timedelta(hours=1)]

print("Number of gaps:", len(gaps))
print("Largest gap:", gaps.max())
```

```python
gaps.value_counts().sort_index()
```

```python
largest_gap = gaps.idxmax()

print("Largest gap ends at:", largest_gap)
print("Gap duration:", gaps.max())
```

```python
df.loc[
    (df['date_time'] >= largest_gap - pd.Timedelta(days=2)) &
    (df['date_time'] <= largest_gap + pd.Timedelta(days=2)),
    ['date_time', 'traffic_volume']
]
```
