# Data Cleaning

This notebook cleans and validates the raw traffic dataset before
feature engineering and model training.

```python
import pandas as pd
df = pd.read_csv(r"/content/Metro_Interstate_Traffic_Volume.csv.gz")
print("Dataset loaded")
```

```python
print("Rows: " ,df.shape[0])
print("Columns: " ,df.shape[1])
```

```python
print("Missing values in each column:")
print(df.isna().sum())
```

```python
missing_percentage = (df.isna().sum() / len(df)) * 100
print(missing_percentage)
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df["date_time"] = pd.to_datetime(df["date_time"])

print("Duplicate timestamps:", df["date_time"].duplicated().sum())
```

```python
print("Rows with zero traffic:", (df["traffic_volume"] == 0).sum())
```

```python
df[df["traffic_volume"] == 0].head(20)
```

```python
df[["temp", "rain_1h", "snow_1h", "clouds_all", "traffic_volume"]].describe()
```

```python
print("Holiday values:")
print(df["holiday"].value_counts(dropna=False))

print("\nWeather main values:")
print(df["weather_main"].value_counts(dropna=False))

print("\nWeather description values:")
print(df["weather_description"].value_counts(dropna=False))
```

```python
print("Negative rain values:", (df["rain_1h"] < 0).sum())
print("Negative snow values:", (df["snow_1h"] < 0).sum())
print("Negative traffic values:", (df["traffic_volume"] < 0).sum())
```

```python
before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Rows before:", before)
print("Rows after:", after)
print("Duplicate rows removed:", before - after)
```

```python
df["holiday"] = df["holiday"].fillna("None")

print(df["holiday"].value_counts())
```

```python
print("Final shape:", df.shape)

print("\nMissing values:")
print(df.isna().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
print("\nFinal data types:")
print(df.dtypes)
```

```python
print("\nNegative values after cleaning:")
print("Negative rain:", (df["rain_1h"] < 0).sum())
print("Negative snow:", (df["snow_1h"] < 0).sum())
print("Negative traffic:", (df["traffic_volume"] < 0).sum())
```

```python
output_path = "/content/traffic_cleaned.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print(output_path)
```
